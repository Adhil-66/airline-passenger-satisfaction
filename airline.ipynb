## Insight with questions

How many rows and columns are present in the airline dataset, and what does each row represent?

```python
df.shape
```

How are passengers distributed across different age groups?

```python
bins = [0, 18, 25, 35, 45, 55, 100]
labels = ['Under 18', '18-25', '26-35',
          '36-45', '46-55', 'Above 55']

df['Age Group'] = pd.cut(
    df['Age'], bins=bins, labels=labels
)

df['Age Group'].value_counts().sort_index()
```

How is flight distance distributed among passengers?

```python
df['Flight Distance'].describe()
```

Which travel class has the highest average flight distance?

```python
df.groupby('Class')['Flight Distance'].mean()
```

How does passenger satisfaction differ between business travel and personal travel?

```python
pd.crosstab(
    df['Type of Travel'],
    df['satisfaction'],
    normalize='index'
) * 100
```

Which five onboard services have the highest average ratings, and which five have the lowest?

```python
service_cols = [
    'Inflight wifi service',
    'Departure/Arrival time convenient',
    'Ease of Online booking',
    'Gate location',
    'Food and drink',
    'Online boarding',
    'Seat comfort',
    'Inflight entertainment',
    'On-board service',
    'Leg room service',
    'Baggage handling',
    'Checkin service',
    'Inflight service',
    'Cleanliness'
]

df[service_cols].mean().sort_values(
    ascending=False
)
```

What are the top five service features with the greatest satisfaction-rating differences between satisfied and dissatisfied passengers?

```python
means = df.groupby('satisfaction')[service_cols].mean()

difference = (
    means.loc['satisfied'] -
    means.loc['neutral or dissatisfied']
).abs()

difference.sort_values(
    ascending=False
).head(5)
```

How does passenger satisfaction vary across different flight-distance ranges?

```python
df['Distance Group'] = pd.cut(
    df['Flight Distance'],
    bins=[0, 500, 1000, 2000, float('inf')],
    labels=['Short', 'Medium', 'Long', 'Very Long']
)

pd.crosstab(
    df['Distance Group'],
    df['satisfaction'],
    normalize='index'
) * 100
```

What is the average departure delay for each customer type?

```python
df.groupby('Customer Type')[
    'Departure Delay in Minutes'
].mean()
```

Which combination of travel class and travel type has the highest satisfaction rate?

```python
pd.crosstab(
    [df['Class'], df['Type of Travel']],
    df['satisfaction'],
    normalize='index'
).mul(100).round(2)
```

Is there a relationship between flight distance and departure delay?

```python
df['Flight Distance'].corr(
    df['Departure Delay in Minutes']
)
```

Is there a relationship between online boarding ratings and passenger satisfaction?

```python
pd.crosstab(
    df['Online boarding'],
    df['satisfaction'],
    normalize='index'
).mul(100).round(2)
```

How many missing values exist in each column, and which column contains the most missing values?

```python
df.isnull().sum().sort_values(
    ascending=False
).head(5)
```

How does filling missing arrival-delay values with the median affect the distribution of arrival delays?

```python
df['Arrival Delay in Minutes'].median()
```

Are there any extreme values in departure delay, arrival delay, or flight distance?

```python
cols = [
    'Departure Delay in Minutes',
    'Arrival Delay in Minutes',
    'Flight Distance'
]

df[cols].describe()
```

Which numerical variables have the strongest positive and negative correlations?

```python
corr = df.select_dtypes(
    include='number'
).corr()

corr
```

Do satisfied and dissatisfied passengers differ in their average ratings for inflight Wi-Fi, seat comfort, and entertainment?

```python
rating_cols = [
    'Inflight wifi service',
    'Seat comfort',
    'Inflight entertainment'
]

df.groupby('satisfaction')[rating_cols].mean().round(2)
```

Which five passenger groups have the lowest satisfaction percentages?

```python
grouped = df.groupby(
    ['Class', 'Type of Travel', 'Customer Type'],
    observed=True
)

rates = grouped['satisfaction'].apply(
    lambda s: s.eq('satisfied').mean() * 100
)

rates.sort_values().head(5).round(2)
```

How do age group, travel class, and type of travel jointly relate to passenger satisfaction?

```python
df['Age Group'] = pd.cut(
    df['Age'],
    bins=[0, 18, 25, 35, 45, 55, 100],
    labels=['Under 18', '18-25', '26-35',
            '36-45', '46-55', 'Above 55']
)

result = pd.crosstab(
    [df['Age Group'], df['Class'], df['Type of Travel']],
    df['satisfaction'],
    normalize='index'
) * 100

result.round(2)
```

- The dataset contains 103,904 passenger records.
- Business class has the highest satisfaction rate at 69.43%.
- Eco class has the lowest satisfaction rate at 18.61%.
- Departure and arrival delays are strongly correlated, with a coefficient of approximately 0.960.
- Neutral or dissatisfied passengers experience higher average departure and arrival delays than satisfied passengers.
- The missing arrival-delay values were handled using median imputation.
Based on these findings, the airline could investigate service quality across travel classes, delay management, and passenger experience. Further service-rating comparisons are needed before deciding which individual service requires the most improvement.
